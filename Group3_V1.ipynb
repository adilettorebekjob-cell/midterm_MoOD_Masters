```python
import sys
import subprocess

with open("results/codet5_gpu_console.log", "w", encoding="utf-8") as log:
    process = subprocess.Popen(
        [sys.executable, "-u", "train.py", "--output", "codet5_gpu"],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        encoding="utf-8",
        errors="replace",
    )
    for line in process.stdout:
        print(line, end="", flush=True)
        log.write(line)
        log.flush()

    returncode = process.wait()

if returncode != 0:
    raise RuntimeError(f"Обучение завершилось с ошибкой: {returncode}")
```
